# Binary classification problem

x1 = hours studied

x2 = hours slept

Predict

0 = Fail

1 = Pass

```python
import numpy as np
import keras
from keras import layers
```

```python
X = np.array([
    [1, 2],
    [2, 2],
    [2, 3],
    [3, 3],
    [4, 3],
    [4, 4],
    [5, 4],
    [5, 5]
], dtype=np.float32)

y = np.array([
    1,
    1,
    1,
    0,
    0,
    0,
    0,
    0
], dtype=np.float32)
```

```python
model = keras.Sequential([
    keras.Input(shape=(2,)),
    layers.Dense(4, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy','recall']
)
history=model.fit(
    X,y,
    epochs=100,
    batch_size=8,
    verbose=1
)
```

```python
model.evaluate(X, y)
```

```python
new_student = np.array([[3, 4]], dtype=np.float32)
prediction=model.predict(new_student)
print(prediction)
probability=prediction[0][0]
if probability < 0.5:
    print("Pass")
else:
    print("Fail")
```
